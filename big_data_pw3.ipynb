```python
import numpy as np
# Сатып алу сомалары
data = np.array([
    5000, 7000, 8000, 6000, 9000,
    7500, 8500, 6500, 10000, 500000
])
# Орташа мән
mean = np.mean(data)
# Стандартты ауытқу
std = np.std(data)
# Z-score есептеу
z_scores = (data - mean) / std
# Аномалияларды анықтау
anomalies = data[np.abs(z_scores) > 2]
print("Орташа мән:", mean)
print("Z-score:", z_scores)
print("Аномалиялар:", anomalies)
```

```python
anomalies= data[np.abs(z_scores)>2]
print('Анықталған аномалиялар', anomalies)
```

```python
indices = np.where(np.abs(z_scores)>2)[0]

print('Аномалия индектері: ', indices)
print('Аномалия мәндері: ', data[indices])
```

```python
clean_data = data[np.abs(z_scores)<=2]

new_mean = np.mean(clean_data)

print('Бастапқы орташа мән: ', np.mean(data))
print('Аномалиясыз орташа мән: ', new_mean)
```

```python
print('Аномалияға дейінгі орташа мән: ', np.mean(data))
print('Аномалиядан кейінгі орташа мән: ', np.mean(clean_data))
```
